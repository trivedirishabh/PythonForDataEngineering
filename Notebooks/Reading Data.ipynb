### Reading Data from Different sources
- Reading "csv"
- Reading "json"
- Reading "parquet"
- Databases using "jdbc"


```python
dbutils.fs.cp("/Workspace/Users/23rishabh2009@gmail.com/DataEngineering/PythonForDataEngineering/Data", dest="/Volumes/pyspark/default/my_volume/", recurse=True)
```

```python
df = spark.read.format("csv")\
    .option("header", True)\
        .option("inferSchema", True)\
            .load("/Volumes/pyspark/default/my_volume/orders.csv")

display(df)
```

```python
df_json= spark.read.format("json")\
    .option("inferSchema", True)\
        .option("multiLine", True)\
            .load("/Volumes/pyspark/default/my_volume/orders.json")

display(df_json)
```

```python
df_parquet = spark.read.format("parquet")\
    .load("/Volumes/pyspark/default/my_volume/part-00000-tid-orders.c000.snappy.parquet")
display(df_parquet)
```

```python
df_jdbc= spark.read.jdbc("jdbc:postgresql://localhost:5432/postgres", "orders", properties={"user": "postgres", "password": "postgres"})
```
