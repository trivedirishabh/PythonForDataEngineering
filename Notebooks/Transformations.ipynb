## Pyspark Transformations
- Simple Select
- Select with col()
- Alias
- filter
    - 'isin' operator
    - '&' operator
    - '|' operator


```python
from pyspark.sql.functions import col
```

```python
df = spark.read.format("csv")\
    .option("header", True)\
        .option("inferSchema", True)\
            .load("/Volumes/pyspark/default/my_volume/orders.csv")
```

```python
df_select = df.select("order_id", "customer_id", "city", "country")
display(df_select)
```

```python
df_select_col = df.select("customer_id", col("gender"), col("country"))
display(df_select_col)
```

```python
df_alias=df.select(col("customer_id"),col("country").alias("customer_country"),col("gender").alias("customer_gender") )
display(df_alias)
```

```python

df_filter = df.filter((col("country") == "India") & ((col("order_status").isin(["Returned","Return to Supplier","Shipped"]))))
display(df_filter)
```
